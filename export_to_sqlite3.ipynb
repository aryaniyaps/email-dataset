## Export existing output to SQLITE3 Database

```python
import pandas as pd
import sqlite3

df = pd.read_csv('output.csv')

# Connect to SQLite database (creates file if it doesn't exist)
conn = sqlite3.connect('export.db')

# table name: 'main'
df.to_sql('main', conn, if_exists='replace', index=False)

# 4. Read Data back from SQLite to verify/visualize
df_from_sql = pd.read_sql('SELECT * FROM main', conn)

# 5. Close connection
conn.close()

# 6. Visualize in IPYNB
df_from_sql
```
